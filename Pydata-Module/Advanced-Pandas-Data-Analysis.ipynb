## Vectorization in Numpy vs Pandas

#### 1. Vectorization of an array of integers using numpy

```python
x = np.array([2, 3, 5, 7, 11, 13])
x * 2
```

#### 2. Vectorization of an array of strings using numpy

```python
# An example of vectorization using strings
data = ['peter', 'Paul', 'MARY', 'gUIDO']
[s.capitalize() for s in data]

# This works but will break with any missing values e.g 'None'
```

#### 3. Vectorization in pandas

```python
import pandas as pd
```

```python
names = pd.Series(data)
names
```

```python
names.str.capitalize()
```

#### Key terms

**Data munging** is the process of cleaning, organizing, and transforming raw, messy data into a clear format that is ready for analysis or use in computer programs.

**Vectorization** in Pandas is a method of performing operations on entire arrays or columns at once, rather than looping through the data row by row.

**NaN** means 'Not a Number'.
