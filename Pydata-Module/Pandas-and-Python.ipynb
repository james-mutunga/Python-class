## Explicit Indexing

```python
dfcopy = df.copy()
```

```python
dfcopy.head()
```

```python
#.columns and .index
dfcopy.columns
```

```python
dfcopy.index
```

```python
# checking for duplicates in the state column and get the number of duplicates in the state column
dfcopy["state"].duplicated().sum()
```

#### Setting a column as the Index

```python
dfindex = dfcopy.set_index("state")
dfindex.head()
```

- Filter the data to only show the details of homeless people in Arizona, Nevada, Texas and Utah

```python
# If we use our original dataframe to do this here is how
df[df["state"].isin(["Arizona", "Texas", "Nevada", "Utah"])]
```

```python
# Subsetting gets easy when the column state is made an index
dfindex.loc[["Arizona", "Texas", "Nevada", "Utah"]]
```

**Index Values don't need to be unique**

- This means you can have a column with duplicate values as the index column

```python
dfcopy["region"].duplicated().sum()
```

```python
dfDuplicateIndex = dfcopy.set_index("region")
dfDuplicateIndex.head()
```

```python
# Subsetting on duplicated index
dfDuplicateIndex.loc["Pacific"]
```

### Multi-level Indexes / Heiratical Indexes

- Using multiple columns as you index

```python
df_ind3 = dfcopy.set_index(["region", "state"])
df_ind3.head()
```

```python
# Zack's way for subsetting df_ind3
idx = pd.IndexSlice
df_ind3.loc[idx[["Pacific", "Mountain"], :], :]
```

**Subsetting the outer level with a list**

```python
df_ind3.loc[["West South Central", "East South Central"]]
```

**Subset Inner levels with a list of tuples**

```python
df_ind3.tail()
```

```python
df_ind3.index
```

```python
# Subsetting with multiindex
df_ind3.loc[[("New England", "Massachusetts"), ("West North Central", "Minnesota")]]
```

### Sorting by Index Values

```python
df_ind3.sort_index()
```

**Controlling Sort Index**

```python
df_ind3.sort_index(level=["region", "state"], ascending=[True, False])

# ascending = True: A-Z
# ascending = False: Z-A
```
