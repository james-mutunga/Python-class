**Creating Arrays From Scratch**

- There are three major ways to create arrays:
- `np.zeros`: This creates an array of zeros
- `np.random.random`: This creates an array with random rumbers
- `np.arrange`: Creates an array with values in a specified range.

```python
# Creating an array with `np.zeros`
np.zeros((5, 3), dtype=int)

# we start with row and the columns.
# shape of the array (5, 3)
# dtype is now `int`

# READ ON DTYPE
```

```python
# Creating an array with `np.random.random`
np.random.random((10, 40))
```

```python
# Creating an Array with np.arrange
np.arange(-3, 4)

# this will give me values from -3 all the way to 4 excluding 4
```

```python
plt.scatter(np.arange(0,7), # We have 7 elements
            np.arange(-3,4)) # We have 7 elements

plt.show()
```

#### Generating Data With Numpy

```python
students_heights = np.round(np.random.normal(1.75, 0.20, 5000), 2)
students_heights

# 2 by 4 array with a mean of 1.75 and a standard deviation (StdDev) of 0.2 and a shape of 5000

# Standard deviation example - imagine a bullseye in archery. 2 players, Bella and Nathan are the archers. Bella's shots are very close to the bullseye (the mean) and Nathan's shots are far from the bullseye. Bella in this case has a lower stand deviation than Nathan.
```

```python
np.std(students_heights)

# this is my standard deviation
```

```python
np.mean(students_heights)

# this is my average
```

```python
students_weights = np.round(np.random.normal(60.32, 15, 5000), 2)
students_weights
```

```python
np_students = np.column_stack((students_heights, students_weights))
np_students
```

```python
np_students.ndim

# this will print the dimensions of the np_students array
```

```python
np.std(np_students[:, 0])

# for all the rows in index 0 - index 0 is the student_heights
# so, we are basically printing the Stddev for student_heights
```
