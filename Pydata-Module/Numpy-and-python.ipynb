**BMI Calculation Illustration**

```python
weight = [45, 60, 70, 65, 67, 70, 40]
```

```python
# Lists do not support Boolean Indexing but Arrays do
# weight > 50
```

```python
height = [130, 160, 170, 200, 210, 140, 159]
```

```python
# Doing this in python gives us a Type error uncomment below to see the error
# weight / height ** 2
# uncomment it to see the error
```

**Numpy Solution**

```python
np_weight = np.array(weight)
```

```python
np_height = np.array(height)
```

```python
bmi = np_weight / np_height ** 2
print(bmi)
```

**Bolean Indexing**

```python
bmi > 0.0020
```

**Creating a 2D Array**

- To make it easy tounderstand view this as two 1D Arrays stack on each other

```python
                #   0      1     2    3      4
np_2d = np.array([[1.75, 1.68, 1.71, 1.89, 1.79],  # 0
                  [65.4, 59.0, 60.9, 70.5, 45.9]]) # 1
np_2d
```

```python
np_2d.shape
```

**Subsetting or indexing in numpy arrays**

```python
np_2d[0]
```

```python
np_2d[0, 2]
```

```python
np_2d[0][2]
```

```python
np_2d[1,:]

# 1 means the row
# : means the column - ':' in python means print all the way to end of that column
```

```python
np_2d[:, 1:3]

# firstly, the ':' tells it to return both rows
# then for both rows return the entries in the 1st and 2nd index because the 3rd one is excluded
```
