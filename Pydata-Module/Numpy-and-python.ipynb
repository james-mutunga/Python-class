```python
np.std(students_heights)

# this is my standard deviation
```

```python
np.mean(students_heights)

# this is my average
```

### Concatinating Numpy Arrays Row Wise

```python
student_ids_scores = np.array([[101, 80], [105, 90], [110, 76], [115, 60]])

student_ids_scores
```

```python
student_ids_scores.shape
```

```python
student_ids_scores.ndim
```

```python
new_students = np.array([[100, 90], [102, 70]])

new_students
```

### Filtering with Fancy Indexing

```python
one_to_five = np.arange(1,6)

# I want to check which of the numbers is divisible by 2

mask = one_to_five % 2 == 0

# I perform the filtering using the mask and I get the numbers that are divisible by 2

one_to_five[mask]
```
