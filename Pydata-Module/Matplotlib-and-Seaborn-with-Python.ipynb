# Reading in Our Data

```python
# Homelessness Data
homelessness = pd.read_csv(r"C:\Users\User\desktop\Python-Class\Data\homelessness.csv")
```

```python
homelessness.head()
```

```python
homelessness.columns
```

```python
# Drop the `Unnamed: 0` column as it is just a duplicate index column
homelessness = homelessness.drop(columns= "Unnamed: 0")
```

```python
homelessness.head()
```

```python
# Create two new columns, `total` and `p_total`
# Which are the sum of the individual and family_member homeless populations
# And the proportion of the total population that is homeless, respectively.

homelessness["total_homeless"] = homelessness["individuals"] + homelessness["family_members"]
homelessness["pct_total_homeless"] = (homelessness["total_homeless"] / homelessness["state_pop"]) * 100
```

```python
homelessness.head()
```

**What Questions can we answer with Visualisations given the data that we have :**
1. What are the Top 5 States in terms of Total homeless individual? (Univariate Chart)
    - Type of Chart : Bar Graphs specifically a horizontal bar graph
    - `y-axis` : state
    - `x-axis` : total homeless

2. Does Population Size Relate to Homelessness?
    - Type of Chart : Combined Bar Chart (Bivariate Chart)
    - `y-axis` : Total Homeless and State Population
    - `x-axis` : State

1. What are the Top 5 states in terms of total homeless individual

```python
# Adding Data to axes

# To the y-axis we will add state
homelessness["state"]

# To the x-axis we will add total_homeless
homelessness["total_homeless"]
```

```python
fig,ax = plt.subplots()
ax.plot(homelessness["state"], homelessness["total_homeless"])
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(homelessness["state"], homelessness["total_homeless"])
#            x-axis                      y-axis
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(homelessness["state"], homelessness["total_homeless"])
#            x-axis                      y-axis
plt.show()
```

```python
# Plot the graph for the top 5 states with the highest total homeless population
top_5_states = homelessness.nlargest(5, 'total_homeless')

fig, ax = plt.subplots()
ax.bar(top_5_states["state"], top_5_states["total_homeless"])
#             x-axis                      y-axis
plt.show()

# This is a Univariate chart in vertical format
```

```python
# Plot the graph for the top 5 states with the highest total homeless population
top_5_states = homelessness.nlargest(5, 'total_homeless')

fig, ax = plt.subplots()
ax.barh(top_5_states["state"], top_5_states["total_homeless"])
#             x-axis                      y-axis
plt.show()

# This is a Univariate chart in horizontal format
```

```python
# Plot the graph for the top 5 states with the highest total homeless population in a horizontal bar chart From the highest to the lowest

# Add a title to the graph and axis labels after colouring each bar in the graph with a different colour
top_5_states = homelessness.nlargest(5, "total_homeless").sort_values("total_homeless")

fig, ax = plt.subplots()
ax.barh(top_5_states["state"], top_5_states["total_homeless"])
ax.set_title("Top 5 States with the Highest Total Homeless Population")
ax.set_xlabel("Total Homeless Population")
ax.set_ylabel("State")
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(top_5_states["state"], top_5_states["total_homeless"])
ax.set_title("Top 5 States with the Highest Total Homeless Population")
ax.set_xlabel("Total Homeless Population")
ax.set_ylabel("State")
plt.show()
```

```python
# Plotting the Seattle and Austin Precipitation Data in Two different plots 

fig, ax = plt.subplots(2, 1, sharey = True)

# Plot the Seattle Precipitation Data in the first plot
ax[0].plot(seattle_weather_sliced["MONTH"], seattle_weather_sliced["MLY-PRCP-NORMAL"], color = "b")
ax[0].plot(seattle_weather_sliced["MONTH"], seattle_weather_sliced["MLY-PRCP-50PCTL"], color = "b", linestyle = "-.")
ax[0].plot(seattle_weather_sliced["MONTH"], seattle_weather_sliced["MLY-PRCP-75PCTL"], color = "b", linestyle = "-.")

# Plot the Austin Precipitation Data in the second plot
ax[1].plot(austin_weather_sliced["MONTH"], austin_weather_sliced["MLY-PRCP-NORMAL"], color = "m")
ax[1].plot(austin_weather_sliced["MONTH"], austin_weather_sliced["MLY-PRCP-50PCTL"], color = "m", linestyle = ":")
ax[1].plot(austin_weather_sliced["MONTH"], austin_weather_sliced["MLY-PRCP-75PCTL"], color = "m", linestyle = ":")

ax[0].set_ylabel("Precipitation (inches)")
ax[1].set_ylabel("Precipitation (inches)")

ax[1].set_xlabel("Time (months)")

plt.show()
```
