# Reading in Our Data

```python
# Homelessness Data
homelessness = pd.read_csv(r"C:\Users\User\desktop\Python-Class\Data\homelessness.csv")
```

```python
homelessness.head()
```

```python
homelessness.columns
```

```python
# Drop the `Unnamed: 0` column as it is just a duplicate index column
homelessness = homelessness.drop(columns= "Unnamed: 0")
```

```python
homelessness.head()
```

```python
# Create two new columns, `total` and `p_total`
# Which are the sum of the individual and family_member homeless populations
# And the proportion of the total population that is homeless, respectively.

homelessness["total_homeless"] = homelessness["individuals"] + homelessness["family_members"]
homelessness["pct_total_homeless"] = (homelessness["total_homeless"] / homelessness["state_pop"]) * 100
```

```python
homelessness.head()
```

```python
# Adding Data to axes

# To the y-axis we will add state
homelessness["state"]

# To the x-axis we will add total_homeless
homelessness["total_homeless"]
```

```python
fig,ax = plt.subplots()
ax.plot(homelessness["state"], homelessness["total_homeless"])
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(homelessness["state"], homelessness["total_homeless"])
#            x-axis                      y-axis
plt.show()
```

```python
fig, ax = plt.subplots()
ax.bar(homelessness["state"], homelessness["total_homeless"])
#            x-axis                      y-axis
plt.show()
```

```python
# Plot the graph for the top 5 states with the highest total homeless population
top_5_states = homelessness.nlargest(5, 'total_homeless')

fig, ax = plt.subplots()
ax.bar(top_5_states["state"], top_5_states["total_homeless"])
#             x-axis                      y-axis
plt.show()

# This is a Univariate chart in vertical format
```

```python
# Plot the graph for the top 5 states with the highest total homeless population
top_5_states = homelessness.nlargest(5, 'total_homeless')

fig, ax = plt.subplots()
ax.barh(top_5_states["state"], top_5_states["total_homeless"])
#             x-axis                      y-axis
plt.show()

# This is a Univariate chart in horizontal format
```

```python
# Plot the graph for the top 5 states with the highest total homeless population in a horizontal bar chart From the highest to the lowest

# Add a title to the graph and axis labels after colouring each bar in the graph with a different colour
top_5_states = homelessness.nlargest(5, "total_homeless").sort_values("total_homeless")

fig, ax = plt.subplots()
ax.barh(top_5_states["state"], top_5_states["total_homeless"])
ax.set_title("Top 5 States with the Highest Total Homeless Population")
ax.set_xlabel("Total Homeless Population")
ax.set_ylabel("State")
plt.show()
```
