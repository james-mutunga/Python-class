```python
import os
from dotenv import load_dotenv
from pymongo import MongoClient

# 1. Load the secret password
load_dotenv()
uri = os.getenv("MONGO_URI")

# 2. Connect to the Client (The Cluster)
client = MongoClient(uri)

# 3. Access the Database (e.g., "school")
# If it doesn't exist, Mongo creates it automatically when you insert data.
db = client["school"]

# 4. Access the Collection (e.g., "students")
students = db["students"]

# 5. Test the connection
try:
    client.admin.command('ping')
    print("Successfully connected to MongoDB!")
except Exception as e:
    print(e)
```

#### Inserting 1 student

```python
new_student = {
    "name": "John Doe",
    "age": 16,
    "gpa": 3.8,
    "courses": ["Math", "Physics"],   # Lists are allowed!
    "address": {"city": "New York"}   # Nested Dicts are allowed!
}

result = students.insert_one(new_student)
print(f"Created Student ID: {result.inserted_id}")
```

#### Inserting many students

```python
batch_data = [
    {"name": "Alice", "age": 14, "gpa": 4.0},
    {"name": "Bob", "age": 15, "gpa": 2.5}
]

result = students.insert_many(batch_data)
print(f"Created {len(result.inserted_ids)} students.")
```

#### Finding one student

```python
# Find Alice
alice = students.find_one({"name": "Alice"})
print(alice)
```

```python
# Find all students with GPA > 3.0
# Syntax: { field: { $operator: value } }
high_achievers = students.find({"gpa": {"$gt": 3.0}})

print("High Achievers:")
for student in high_achievers:
    print(f"- {student['name']} (GPA: {student['gpa']})")
```

#### Editing/Updating one student's query (age)

```python
# Change John's age to 17
students.update_one(
    {"name": "John Doe"},       # Filter (Who?)
    {"$set": {"age": 17}}       # Action (What?)
)
```

```python
students.update_one(
    {"name": "John Doe"},
    {"$push": {"courses": "Chemistry"}}
)
```

```python
# It's John's birthday! Add 1 to his age.
students.update_one(
    {"name": "John Doe"},
    {"$inc": {"age": 1}}
)
```

#### Deleting one student

```python
students.delete_one({"name": "Bob"})
```

#### Deleting many students

```python
# Delete everyone with a low GPA
result = students.delete_many({"gpa": {"$lt": 2.0}})
print(f"Expelled {result.deleted_count} students.")
```
