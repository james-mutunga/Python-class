**Creating Our First Class Customer**

```python
class CustomerGikosh:
    """A class that will be a blueprint for all the customers in Gikosh
    It will contain the following states and behaviours:
    - States: name, rewards_number, email, phone_number
    - Behaviours: make_purcase, return_purchase, order_online, cancel_order"""

    # "__init__" is a dunder function that we use to initialize the states for the customer class. This instantiates the states of the code. The double underscore is what makes it a dunder function.
    # self here will allow as to add multiple name, reward numbers, emails and phone numbers
    # this way every customer will have their own name, rewards_number, email, phone_number

    # Here the name, rewards_number etc. all states are our `attributes`
    # Our make_purchase, return_purchase, order_online etc. are our `methods`

    def __init__(self, name, rewards_number, email, phone_number):
        self.name = name
        self.rewards_number = rewards_number
        self.email = email
        self.phone_number = phone_number

    def make_purchase(self):
        print(f"Thank you {self.name} for your purchase. You have earned 200 points under reward number {self.rewards_number}")

    def return_purchase(self):
        print("Return of Purchase for order 092 is approved. You will recieve a refund in 3-5 business days.")

    def order_online(self):
        print(f"Thank you for your online order {self.name}.")

    def cancel_order(self):
        print(f"Your order for {self.name} has been cancelled.")
```

```python
cust0ne = CustomerGikosh("James", 1010, "mutungajames334@gmail.com", "254717721596")
```

```python
# View all our customer one details
print(cust0ne.name)
print(cust0ne.rewards_number)
```

```python
cust0ne.return_purchase()
```

```python
class MPESAcustomer:

    """A class that will be a blueprint for all MPESA customers
        It will contain the following states and behaviours:
        - States: name, phone_number, account_balance, fuliza_limit
        - Behaviours: check_account_balance, buy_goods, send_money_request, check_fuliza_limit"""

    def __init__(self, name, phone_number, account_balance, fuliza_limit):
        self.name = name
        self.phone_number = phone_number
        self.account_balance = account_balance
        self.fuliza_limit = fuliza_limit

    def send_money_request(self):
        amount = input("Enter Amount You would like to request: ")
        print(f"Thank you {self.name} for requesting {amount}KES from Patrick Kimani.")

    def buy_goods(self):
        goods_amount = input("Enter Amount You would like to send: ")
        print(f"Sending {goods_amount}KES to Kipkemboi's barber shop.")

    def check_account_balance(self):
        print(f"Dear {self.name}, Your current MPESA Balance is {self.account_balance}KES for {self.phone_number}")
                
    def check_fuliza_limit(self):
        print(f"Dear {self.name}, your current outstanding fuliza limit is {self.fuliza_limit}KES")
```

```python
class Animal:
    """This is a an animal class and it contains the following states and behaviours
    States: name
    Behaviours: speak"""

    def __int__(self, name):
        self.name = name

    def speak(self):
        print("This animal makes a sound")

class Dog:
    """This class dog contains the following states and behaviours
    States: name
    Behaviours: speak"""

    def speak(self):
        print("Dog barks!")

class Cat(Animal):
    """This cat class inherits from the animal class this means it contains 
    both the states from the animal class and states that we will define inside it
    States: name
    Behaviour: speak"""

    def speak(self):
        print("Cat meows!")


animals = [Dog(), Cat()]
for animal in animals:
    animal.speak()
```
