> 1. Replacing Parts of a string : " My Fave Colour is red" replace red with your actual fave colour 

```python
colour = "Red"
print(f"My Fave colour is {colour}")
```

```python
colour = colour.replace("Red" , "Green")

#                 ^ 
#       This replaces Red 
#           with Green
#                         ^         ^
#                       old        new

print(f"My Fave colour is {colour}")

# The replace keyword allows for 2 values
```

> 2. Changing the case of strings. "My name is Ladorna" - Change Lardona into fully Upper and Fully lower case

```python
name = "My name is Lardona"
print(name)
```

```python
new_name = name.upper()
print(new_name)
```
