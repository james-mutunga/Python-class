#### 2. Analyzing property prices

```python
data[:5]
```

```python
prices = data['Price']

print("First 5 prices:", prices[:5])
print("Data type of prices:", prices.dtype)
```

#### 3. Analysing property size

```python
Average_price = np.mean(prices)
print(f"Average price is {Average_price}")

Median_price = np.median(prices)
print(f"Median price is {Median_price}")

Standard_deviation = np.std(prices)
print(f"Standard deviation is {Standard_deviation}")

Minimum_price_value = np.min(prices)
print(f"Minimum price value is {Minimum_price_value}")

Maximum_price_value = np.max(prices)
print(f"Maximum price value is {Maximum_price_value}")
```

#### 4. Calculating price per square foot

```python
sizes = data['Square_Feet']

price_per_sqft = prices / sizes
```

```python
Average_price_per_sqft = np.mean(price_per_sqft)
print(f"The average price per square foot is {Average_price_per_sqft}")

Median_price_per_sqft = np.median(price_per_sqft)
print(f"The median price per square foot is {Median_price_per_sqft}")

Minimum_price_per_sqft = np.min(price_per_sqft)
print(f"The minimum price per square foot is {Minimum_price_per_sqft}")

Maximum_price_per_sqft = np.max(price_per_sqft)
print(f"The maximum price per square foot is {Maximum_price_per_sqft}")
```

#### 5. Iding expensive properties

```python
# Finding the 75th Percentile Threshold of the prices of each housing asset in the dataset

price_75th = np.percentile(prices, 75)
```

```python
expensive_mask = prices >= price_75th
```

```python
np.sum(expensive_mask)

# This result is the number of properties that are in the `espensive_mask`; properties that are in the top 25% by price
```

```python
np.mean(prices[expensive_mask])
```

```python
expensive_mean = np.mean(prices[expensive_mask])
```

```python
print("Average price of top 25% properties:", expensive_mean)
```

#### 6. Applying multiple conditions

```python
# Get thresholds
avg_size = np.mean(sizes)
price_75th = np.percentile(prices, 75)

# Combine the 3 conditions using () and &
multi_mask = (sizes > avg_size) & (prices >= price_75th) & (data['Location_Score'] >= 8)

# Count how many properties meet all 3 criteria
count = np.sum(multi_mask)
print("Number of properties meeting all 3 criteria:", count)
```

#### 7. Comparing prices to the average

```python
# 1. Get the price
prices = data['Price']

# 2. Calculate the average price
overall_mean = np.mean(prices)

# 3. Perform broadcasting: subtract average from all prices
price_diff = prices - overall_mean

# 4. Count properties above and below average
above_avg_count = np.sum(prices > overall_mean)
below_avg_count = np.sum(prices < overall_mean)

print("Overall Mean Price:", overall_mean)
print("Properties Above Average:", above_avg_count)
print("Properties Below Average:", below_avg_count)
print("Total Properties Checked:", above_avg_count + below_avg_count)
```

> Broadcasting made it possible to subtract the single number overall_mean from all items in the prices array at once

#### 8. Find the 10 most expensive properties

```python
top_10_indices = np.argsort(prices)[-10:]

# This sorts the array from lowest to highest and returns their position index numbers. 
# Taking [-10:] grabs the indices of the highest 10 values:
```

```python
top_10_properties = data[top_10_indices]

print(top_10_properties)
```

#### 9. Final Analysis & Business Insight

```python
# Comparing Top 10 against Overall Dataset

# Top 10 average prices against overall average prices
print("Top 10 Average Price:", np.mean(top_10_properties['Price']))
print(f"Overall Average Price: {np.mean(prices)}\n")

# Top 10 average sizes against overall average size
print("Top 10 Average Size:", np.mean(top_10_properties['Square_Feet']))
print(f"Overall Average Size: {np.mean(sizes)}\n")

# top 10 average number of bedrooms against overall number of average bedrooms
print("Top 10 Average Bedrooms:", np.mean(top_10_properties['Num_Bedrooms']))
print(f"Overall Average Bedrooms: {np.mean(data['Num_Bedrooms'])}\n")

# top 10 average location score of against overall locaton score
print("Top 10 Average Location Score:", np.mean(top_10_properties['Location_Score']))
print(f"Overall Average Location Score: {np.mean(data['Location_Score'])}\n")
```

> Price, Size, Number of Bedrooms, Location score

- The top 10 properties are more expensive than average by a difference of almost $300,000. They're also signifantly larger than the average size by a difference of 93 square feet. The top 10 properties have more than 4 bedrooms compared to the average of 2+ bedrooms on average. They also have a higher location score overall.
