```python
def get_weather(city):
    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={API_KEY}&units=metric"
    # This is the url of our response
    response = requests.get(url)
    # This is our response definition which 
    if response.status_code == 200:
    # e.g error 404 is a response.status_code for an error, a respons.status_code of 200 indicates success
        data = response.json()
        main = data['main']
        temperature = main['temp']
        weather_description = data['weather'][0]['description']
        return f"The temperature in {city} is {temperature}°C with {weather_description}."
    else:
        return "I couldn't retrieve the weather information."
```

```python
import requests

# Free, no API key needed, supports KES and 160+ other currencies: https://open.er-api.com
def convert_currency(amount, from_currency, to_currency):
    url = f"https://open.er-api.com/v6/latest/{from_currency}"
    response = requests.get(url)

    if response.status_code == 200:
        data = response.json()
        rate = data["rates"][to_currency]
        converted_amount = amount * rate
        return f"{amount} {from_currency} is equal to {round(converted_amount, 2)} {to_currency}"
    else:
        return "I couldn't retrieve the exchange rate information."

print(convert_currency(5000, "EUR", "KES"))
```

**Lambda Functions**

- Are a quick and easy way to write functions with the keyword `lambda`

```python
def raisea2b(a, b):
    """A function that takes two parameters a and b and raises a to the power of b"""
    raised_to_b = a ** b
    print(raised_to_b)

raisea2b(20, 3)
```

```python
raisex2y_short = lambda x, y: x ** y
print(raisex2y_short(20, 3))

# function
# def           function
#                keyword

# When you're defining a lambda function you start with the name of the function, in this case "raisex2y_short"
# Then you insert the function keyword `lambda` and input your values, in this case x and y.
# Then we insert the lamda function definition, in this case x ** y (x raised to the power of y)
```

## Regular Expressions

- A Regular Expression is a string Containing a combination of `normal characters` and `special characters` that describes **patterns** to find text or positions witihin a text

```python
import re
```

```python
movie_str = "I really enjoyed watching the Movie '#1Spid7er Man: Brand New Day' and I would recommend it"
```

```python
# Use Regular Expressions to only find the movie name in the string `movie_str` above

"""
The Syntax of Find all is :
're.findall(r'regex_pattern', string_to_search)
"""

re.findall(r"#1Spid\d\w\w\s", movie_str)

#`Spid\d` will not find anything because .findall is looking for that specific pattern.
# `Spid\d` will scan and stop at `Spid\d` because `Spid\der` has an `er' at the end. 
# But if we were to add a space in between `#1Spid 7er` then it would print.
# Why? because it matches the following pattern.
# If we were to input '#1Spid\w\w\s' it would then print '#1Spid7er' because it follows the same parttern. 'digit', 'word', 'word', 'space'.
```

| Metacharater| Meaning     |Code Example| What the Code Does|
|------|------|------|------|
|   \d   | Digit     |  `re.findall(r'User\d', "The Winners for the Hackathon are User6 and User8")`|Finds all characters with the word User followed by a digit| 
|   \D  | Non-Digit     |  `re.findall(r'User\D', "The Winners for the Hackathon are UserK, UserL and User8")`|Finds all characters with the word User followed by a non-digit| 
|   \w  | word  |  `re.findall(r'User\w', "The Winners for the Hackathon are User9, UserJ and UserPQ")`|Finds all characters with the word User followed by any digit or normal character|
|   \W  | Non-word  |  `re.findall(r'\W\d\d', "The Concert tickets for Fally Ipupa's concert were $47")`|Finds a price inside our string. We use `\W` to match the dollar sign followed by two digits|
|   \s  | Whitespace  |  `re.findall(r'Zindua\sSchool', "I am having a great time learning at Zindua School")`|Specifies a Whitespace|
|   \S  | Non-Whitespace  |  `re.sub(r'Mashujaa\SDay','Mashujaa Day', "In the month of October we celebrate Mashujaa-Day In Kenya")`|Detects matches of `Mashujaa folowed by any non-space character followed by Day and replace them with the word 'Mashujaa Day'`|

**Sample \d Metacharacter**

```python
winners = "The Winners for the Hackathon are User6 and User8"

re.findall(r'User\d', winners)
```

**Sample \D Metacharacter**

```python
winners2 = "The Winners for the Hackathon are UserK, UserL and User8"

re.findall(r'User\D', winners2)

# This expression will only print the Users that follow the pattern User\D
```

**Sample \w Metacharacter**

```python
winners3 = "The Winners for the Hackathon are User9, UserJ and UserPQ"

re.findall(r'User\w' or r'User\w\w', winners3)

# In the case of `and` we will only get the entries with both conditions that have been met.
# In the case of `or` we will get both entries that match user\w or 'user\w\w'.
```

**Sample \W Metacharacter**

```python
re.findall(r'\W\d\d', "The Concert tickets for Fally Ipupa's concert were $47")

# In the case of 'W\d\d' we will have the result as '$47'
# Why? because '$47' is the only entry that followed the pattern 'Non-Word followed by two digits'
```

**Sample \s Metacharacter**

```python
re.findall(r'Zindua\sSchool', "I am having a great time learning at Zindua School")
```

**Sample \S Metacharacter**

```python
re.sub(r'Mashujaa\SDay','Mashujaa Day', "In the month of October we celebrate Mashujaa-Day In Kenya")

# Here re.sub means 'substitue'.
# What it does is it reads through the sentence and finds all the entries with the pattern 'Mashujaa' followed by a Non-Whitespace then 'Day'
# Then once it has been foud it is then replaced by 'Mashujaa Day'

# So, in short, the `.sub` function; which takes in two entries '' '', replaces the leftmost entry with the one at the end in the sentence above
```

## Web Scraping

- Web scraping is the process of using certain elements from a website and including it in our code.

- To avoid legal issues we need to confirm what each website allows us to use by searching 'www.website.com/robots.txt' for example: 'tuko.com/robots.txt'

```python
import webbrowser
import requests
import threading
from bs4 import BeautifulSoup
```

```python
"""
Below is a function that we can use to scrape news from CNN
"""

def scrape_news():
    url = 'https://edition.cnn.com/world'
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.content, 'html.parser')
        
        # CNN uses different tags for headlines
        headlines = soup.find_all('span', class_='container__headline-text')  # Adjust according to CNN's structure

        # Print the number of headlines found for debugging
        print(f"Found {len(headlines)} headlines.")
        
        news_list = [headline.get_text().strip() for headline in headlines[:5]]  # Get the top 5 headlines
        if news_list:
            return "\n".join(news_list)
        else:
            return "No headlines found."
    else:
        return "Failed to retrieve news."

scrape_news()
```
