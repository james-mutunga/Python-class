```python
attained_grade = 10
target_grade = 50

# Comparing of the grade

if attained_grade >= target_grade:
    print("Target Grade Achieved")
    
else:
    print("Target Grade Not Achieved.")
```

### Accessing Elements in a List

```python
# I want to access and print 20, 11 and 30 from the list below

prices = [10, 20, 11, 30, 40]

#          0   1   2   3   4            index numbers
#         -5  -4  -3  -2  -1

prices[1:4]

#      1                    is the first index of the list we want to access
#        4                  excludes the last index of the list we want to access. It will print up to index 3 which is 30

```

```python
prices[2:5]
```

```python
# we want access to 10, 11, 40

# prices[0], prices[2], prices[4]

prices[0:5:2]

#      0 5 2

#      0 - will print the first index of the list we want to access
#        5 - will exclude the last index of the list we want to access
#          2 - indicates the step size we want to take within the list

#result [10, 11, 40]
```

### Sets and Tuples

- Contain Unique Data
- Unchangable:
    - You can add or remove values but you cannot change them
- Ideal to identify and remove duplicates
- It is quick to search compared to other data structures such as lists

**Creating a Set**

- Set = {}
- Sets do not include ':' = Dictionaries
- No ':' = Set

```python
mahamri_ingredients = {"coconut milk", "cardamon", "sugar", 
                       "wheat flour", "yeast", "fresh palm wine", 
                       "cooking oil"}

print(mahamri_ingredients)
type(mahamri_ingredients)
```

**Converting to a Set**

```python
duplicate_ingredients = ["coconut milk","cardamon", "sugar", "wheat flour" "yeast",  "cooking oil", "fresh palm wine", "cooking oil","cardamon", "sugar"]

print(type(duplicate_ingredients))

# This is a list
```

```python
unique_ingredients = set(duplicate_ingredients)

# Here we have converted the "duplicate ingredients" to a set
```

```python
print(unique_ingredients)
type(unique_ingredients)
```

**Limitations of Sets:**
- Sets do not have an index meaning they are not ordered
    - Can't subset sets [] / slice them
    - We cannot then have duplicates
    - Sets are mutable

### Loops

- We have the following types of Loops in Python

```python
student_grades = [10, 13, 65, 59, 21, 45, 90, 33, 0, 88]
```

```python
student_grades[2] >= 50
```

```python
print(student_grades)
```

```python
# Printing the grade of each student in the list
for grade in student_grades:
    print(grade)
```

**Conditional Statements in For Loops**

```python
for grade in student_grades:
    # Check if the grade is greater than or equal to 50
    if grade >= 50:
        print(grade)
```

```python
print(student_grades)
```

```python
# Check how many students have scored 50 and above
graduating_students = 0
students_not_graduating = 0

# At the start there are 0 graduating students and 0 students not graduating

for grade in student_grades:
        # Check if the grade is greater than or equal to 50
        if grade >= 50:
                graduating_students += 1 # This is an increment operator
                print(graduating_students)
        else:
                students_not_graduating += 1

print(f"Number of students who are graduating: {graduating_students}")
print(f"Number of students who are not graduating: {students_not_graduating}")
print(f"The Total Number of Students: {graduating_students + students_not_graduating}")
```

### While loops

> `break` and `continue` statement

- While loops are very useful for continous tasks :
    - Accelerate `while` a button is being pressed
    - Monitor `while` below or above threshold

```python
# Items In Stock
stock = 10

# Number of Purchases
purchases = 0

while purchases < stock:
    # When a customer buys something increase the number of purchases by 1
    purchases += 1

    # Print for me remaining stock after each purchase
    print(f"Remaining stock: {stock - purchases}")
```

### Sample Use of Break and Continue

```python
# Items In Stock
stock = 10

# Number of Purchases
purchases = 0

while purchases < stock:
    # When a customer buys something increase the number of purchases by 1
    # purchases += 1

    # Print for me remaining stock after each purchase
    print(f"Remaining stock: {stock - purchases}")

    # Terminate the LOOP
    break
```
